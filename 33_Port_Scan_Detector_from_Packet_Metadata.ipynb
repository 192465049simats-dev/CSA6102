```python
 import pandas as pd
from google.colab import files

print("Upload Packet Metadata CSV File")
uploaded = files.upload()

filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

PORT_THRESHOLD = 10

print("\n" + "="*75)
print("           PORT SCAN DETECTOR FROM PACKET METADATA")
print("="*75)

port_counts = df.groupby("SourceIP")["DestinationPort"].nunique().reset_index()
port_counts.rename(columns={"DestinationPort": "UniquePortsScanned"}, inplace=True)

alerts = []

for _, row in port_counts.iterrows():
    if row["UniquePortsScanned"] >= PORT_THRESHOLD:

        scan_data = df[df["SourceIP"] == row["SourceIP"]]

        alerts.append({
            "Source IP": row["SourceIP"],
            "Ports Scanned": row["UniquePortsScanned"],
            "Packets Sent": len(scan_data),
            "Protocol": scan_data.iloc[0]["Protocol"],
            "Risk": "HIGH"
        })

if alerts:
    result = pd.DataFrame(alerts)
    print(result.to_string(index=False))
else:
    print("No Port Scanning Activity Detected.")

print("\nAnalysis Completed Successfully.")
print("="*75)
```
