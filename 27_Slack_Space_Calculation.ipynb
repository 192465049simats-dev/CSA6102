```python


print("=" * 55)
print("        SLACK SPACE CALCULATION")
print("=" * 55)

while True:

    print("\nMenu")
    print("1. Calculate Slack Space")
    print("2. Exit")

    choice = input("\nEnter your choice: ")

    if choice == "1":

        file_size = int(input("Enter File Size (bytes): "))
        cluster_size = int(input("Enter Cluster Size (bytes): "))

        if cluster_size <= 0:
            print("Cluster size must be greater than zero.")
            continue

        remainder = file_size % cluster_size

        if remainder == 0:
            slack_space = 0
            clusters = file_size // cluster_size
        else:
            slack_space = cluster_size - remainder
            clusters = (file_size // cluster_size) + 1

        allocated_space = clusters * cluster_size

        print("\n" + "-" * 45)
        print("RESULT")
        print("-" * 45)
        print("File Size           :", file_size, "bytes")
        print("Cluster Size        :", cluster_size, "bytes")
        print("Clusters Required   :", clusters)
        print("Allocated Space     :", allocated_space, "bytes")
        print("Slack Space         :", slack_space, "bytes")
        print("-" * 45)

    elif choice == "2":
        print("\nProgram Closed.")
        break

    else:
        print("Invalid choice! Please try again.")
```
