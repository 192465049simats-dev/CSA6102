```python
 import pandas as pd
from google.colab import files

print("Upload USB Device History CSV File")
uploaded = files.upload()

filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

print("\n========== USB DEVICE HISTORY ==========")
print(df)

print("\n===================================================")
print(" WINDOWS REGISTRY USB DEVICE HISTORY ANALYZER")
print("===================================================")

print("\nTotal USB Devices Detected :", len(df))

print("\nUSB Device Details\n")

for index, row in df.iterrows():

    print("-------------------------------------------")
    print("Device Name      :", row["DeviceName"])
    print("Vendor           :", row["Vendor"])
    print("Product ID       :", row["ProductID"])
    print("Serial Number    :", row["SerialNumber"])
    print("First Connected  :", row["FirstConnected"])
    print("Last Connected   :", row["LastConnected"])

print("\n===================================================")

print("\nMost Recently Connected USB Device")

latest = df.sort_values("LastConnected", ascending=False).iloc[0]

print("-------------------------------------------")
print("Device Name :", latest["DeviceName"])
print("Vendor      :", latest["Vendor"])
print("Last Used   :", latest["LastConnected"])

print("\n===================================================")

print("Analysis Completed Successfully.")
```
