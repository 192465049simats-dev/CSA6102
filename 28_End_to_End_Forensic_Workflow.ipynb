```python

import hashlib
import datetime
from google.colab import files

print("=" * 65)
print("      END-TO-END DIGITAL FORENSIC WORKFLOW")
print("=" * 65)
print("\nSTEP 1 : Upload Digital Evidence")
uploaded = files.upload()

filename = list(uploaded.keys())[0]

print("\nEvidence Collected Successfully")
print("Evidence File :", filename)

print("\nSTEP 2 : Calculating SHA-256 Hash")

sha256 = hashlib.sha256()

with open(filename, "rb") as f:
    while True:
        data = f.read(4096)
        if not data:
            break
        sha256.update(data)

original_hash = sha256.hexdigest()

print("SHA-256 Hash")
print(original_hash)

print("\nSTEP 3 : Integrity Verification")

verify = input("Enter SHA-256 hash for verification: ")

if verify.lower() == original_hash.lower():
    integrity = "PASSED"
    print("Integrity Verification : PASSED")
else:
    integrity = "FAILED"
    print("Integrity Verification : FAILED")

print("\nSTEP 4 : Evidence Analysis")

import os

size = os.path.getsize(filename)

print("File Name :", filename)
print("File Size :", size, "bytes")

extension = filename.split(".")[-1]

print("File Type :", extension)

print("\nSTEP 5 : Chain of Custody")

investigator = input("Enter Investigator Name: ")

current_time = datetime.datetime.now()

print("Investigator :", investigator)
print("Date & Time :", current_time)

print("\nSTEP 6 : Generating Report")

report = f"""
==============================
 DIGITAL FORENSIC REPORT
==============================

Evidence File
-------------
{filename}

File Size
---------
{size} bytes

File Type
---------
{extension}

SHA-256
--------
{original_hash}

Integrity Status
----------------
{integrity}

Investigator
------------
{investigator}

Collection Time
---------------
{current_time}

Conclusion
----------
Evidence processed successfully.

"""

report_name = "Forensic_Report.txt"

with open(report_name, "w") as f:
    f.write(report)

print("\nReport Generated Successfully")
print("Report File :", report_name)

print("\nWorkflow Completed Successfully.")
```
