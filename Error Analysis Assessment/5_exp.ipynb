```python
import hashlib

def compute_hash(filepath):
    hasher = hashlib.sha256()
    with open(filepath, 'r') as f:
        content = f.read()
    hasher.update(content)
    return hasher.digest

def verify_integrity(filepath):
    hash_at_collection = compute_hash(filepath)
    hash_before_analysis = compute_hash(filepath)
    if hash_at_collection == hash_before_analysis:
        print('Match: Evidence integrity verified')
    else
        print('Mismatch: Evidence may be compromised')

verify_integrity('evidence_image.dd')
```

```python
 import hashlib
from google.colab import files

print("Upload the evidence file:")
uploaded = files.upload()

filepath = list(uploaded.keys())[0]

def compute_hash(filepath):
    hasher = hashlib.sha256()

    with open(filepath, "rb") as f:
        while True:
            chunk = f.read(4096)
            if not chunk:
                break
            hasher.update(chunk)

    return hasher.hexdigest()

def verify_integrity(filepath):

    hash_at_collection = compute_hash(filepath)

    hash_before_analysis = compute_hash(filepath)

    print("\nHash at Collection : ", hash_at_collection)
    print("Hash Before Analysis:", hash_before_analysis)

    if hash_at_collection == hash_before_analysis:
        print("\nMatch: Evidence integrity verified")
    else:
        print("\nMismatch: Evidence may be compromised")

verify_integrity(filepath)
```
