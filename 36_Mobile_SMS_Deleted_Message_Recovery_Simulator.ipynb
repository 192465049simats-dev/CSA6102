```python

import pandas as pd
from google.colab import files

print("Upload SMS Database CSV File")
uploaded = files.upload()

filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

print("\n" + "="*75)
print("         MOBILE SMS DELETED-MESSAGE RECOVERY SIMULATOR")
print("="*75)
deleted_messages = df[df["Status"] == "Deleted"]

if deleted_messages.empty:
    print("\nNo Deleted Messages Found.")
else:
    print("\nRecovered Deleted Messages:\n")
    print(deleted_messages.to_string(index=False))

print("\n")
print("="*75)
print("Total Messages           :", len(df))
print("Recovered Deleted SMS    :", len(deleted_messages))
print("="*75)
print("Recovery Simulation Completed Successfully.")
```
